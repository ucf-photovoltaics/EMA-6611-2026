```python
# Task 4 validation — upload survey.csv to this Colab session, then run this cell.
import pandas as pd

REQUIRED = ["first_author", "year", "title", "doi", "device_type", "absorber_emitter",
            "key_fom_name", "key_fom_value", "processing_disclosed", "top20", "bibtex_key"]

df = pd.read_csv("survey.csv")
missing_cols = [c for c in REQUIRED if c not in df.columns]
assert not missing_cols, f"Missing required columns: {missing_cols}"

n = len(df)
n_top = (df["top20"].astype(str).str.strip().str.lower() == "yes").sum()
n_proc = (df["processing_disclosed"].astype(str).str.strip().str.lower() == "yes").sum()
empties = {c: int(df[c].isna().sum()) for c in REQUIRED if df[c].isna().sum() > 0}

print(f"Articles in corpus:        {n}  (need >= 50)")
print(f"Marked top20 == yes:       {n_top}  (need exactly 20)")
print(f"processing_disclosed yes:  {n_proc}  (need {n})")
print(f"Columns with empty cells:  {empties if empties else 'none'}")
print(f"Year range:                {int(df['year'].min())}-{int(df['year'].max())}")
print()
assert n >= 50, "Corpus has fewer than 50 articles."
assert n_top == 20, "Exactly 20 rows must have top20 == yes."
assert n_proc == n, "Every row must have processing_disclosed == yes."
print("Validation PASSED. The summary below is part of your submission:")
df.groupby("year").size().rename("articles_per_year").to_frame().T
```
