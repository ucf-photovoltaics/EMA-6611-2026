### Background: what a contact has to do

Homework 2 treated the carrier-selective junction — the layers that decide *which* carrier leaves through each side. Metallization is the next step outward: the metal that touches those layers and the conductors that carry the current to the terminals. It has to make a low-resistance (ohmic) contact, conduct laterally without wasting power, cover as little of the optical aperture as the device allows, stick, and survive the rest of the process. Three ideas organize the subject.

**The barrier.** In the ideal Schottky–Mott picture a metal of work function $W_m$ on a semiconductor of electron affinity $\chi$ gives barriers $\Phi_{Bn} = W_m - \chi$ for electrons and $\Phi_{Bp} = E_g - \Phi_{Bn}$ for holes. Real interfaces are pinned by interface states, so the measured barrier depends only weakly on the metal (Sze & Ng, *Physics of Semiconductor Devices*, 3rd ed. (2007), Ch. 3; Grundmann, *The Physics of Semiconductors*, 4th ed., Section 21.2).

**Getting through it.** Carriers cross a barrier over the top (thermionic emission, TE), through it (field emission, FE) or part-way up and through (thermionic-field emission, TFE). The regime is set by the ratio of the tunneling energy $E_{00} = (q\hbar/2)\sqrt{N/(m^{*}\varepsilon_s)}$ to $kT$ (Padovani & Stratton, Solid-State Electron. 9, 695 (1966)), and the figure of merit is the **specific contact resistivity** $\rho_c = (\partial J/\partial V)^{-1}_{V=0}$ in $\Omega$ cm$^2$. A compact interpolation across the three regimes is

$$\rho_c \approx \frac{k}{qA^{*}T}\exp\!\left(\frac{\Phi_B}{E_0}\right), \qquad E_0 = E_{00}\coth\!\left(\frac{E_{00}}{kT}\right)$$

(Yu, Solid-State Electron. 13, 239 (1970); Schroder, *Semiconductor Material and Device Characterization*, 3rd ed. (2006), Ch. 3): at low doping $\rho_c$ depends only on the barrier; at high doping it falls steeply with the doping under the metal. That is why ohmic contacts are made by heavy doping under the metal, by alloying or annealing a reactive metal stack, or by a degenerate interlayer such as a transparent conductive oxide (TCO).

**The cost at device level.** The same milliohms matter differently in each device. In a solar cell, contact, finger and sheet resistances add to a series resistance $R_S$ that costs fill factor, and the grid is a trade against shading (Meier & Schroder, IEEE Trans. Electron Devices 31, 647 (1984)). In an LED the top contact is opaque, so the question is how far the current spreads beyond it, $L_s = \sqrt{t\,n_{ideal}kT/(q\rho J_0)}$ (Thompson, *Physics of Semiconductor Laser Devices* (1980); Schubert, *Light-Emitting Diodes*, 3rd ed. (2018)), and $R_S$ appears as excess forward voltage. In a photodetector the contact resistance and pad capacitance enter the $RC$ bandwidth.

```python
# ============================================================
# HW04 MODULE A — metallization calculators: barriers, contact resistivity, TLM, grids, series resistance,
# transparent conductors, current spreading, photodiode bandwidth
# (instructor-provided; run, do not modify)
# ============================================================
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

kT = 0.025852            # eV at 300 K (= thermal voltage in V)
q_C = 1.602176634e-19    # C
eps0 = 8.8541878128e-14  # F/cm
hbar = 1.054571817e-34   # J s
m0 = 9.1093837015e-31    # kg
kB_eV = 8.617333e-5      # eV/K

# Semiconductor parameters for contact calculations at 300 K. Starting points only — replace with values from your
# corpus and cite them. eps_r: static relative permittivity; m: tunneling effective mass (units of m0); Astar: effective
# Richardson constant (A cm^-2 K^-2; 120*m if not measured); chi: electron affinity (eV); Eg: band gap (eV).
# Sources: Sze & Ng (2007), Ch. 1 and 3 and App. F–G; Levinshtein et al. (1996); Schubert (2018) for GaN.
CONTACT = {
    'n-Si':   dict(eps_r=11.7, m=0.26,  Astar=112.0, chi=4.05, Eg=1.12),
    'p-Si':   dict(eps_r=11.7, m=0.39,  Astar=32.0,  chi=4.05, Eg=1.12),
    'n-GaAs': dict(eps_r=12.9, m=0.067, Astar=8.2,   chi=4.07, Eg=1.42),
    'p-GaAs': dict(eps_r=12.9, m=0.50,  Astar=74.0,  chi=4.07, Eg=1.42),
    'n-InP':  dict(eps_r=12.5, m=0.08,  Astar=9.4,   chi=4.38, Eg=1.34),
    'n-GaN':  dict(eps_r=8.9,  m=0.20,  Astar=26.4,  chi=4.10, Eg=3.40),
    'p-GaN':  dict(eps_r=8.9,  m=0.80,  Astar=96.0,  chi=4.10, Eg=3.40),
    'p-CdTe': dict(eps_r=10.2, m=0.40,  Astar=48.0,  chi=4.28, Eg=1.50),
}
# Work functions of contact metals (eV), polycrystalline values after Michaelson, J. Appl. Phys. 48, 4729 (1977).
WORK_FUNCTION = {'Mg': 3.66, 'In': 4.12, 'Ag': 4.26, 'Al': 4.28, 'Ti': 4.33, 'Cr': 4.50, 'Mo': 4.60, 'Cu': 4.65,
                 'Au': 5.10, 'Pd': 5.12, 'Ni': 5.15, 'Pt': 5.65}


def _semi(semiconductor):
    """Accepts a CONTACT key or a dict with your own cited values (label, eps_r, m, and optionally Astar, chi, Eg)."""
    if isinstance(semiconductor, dict):
        p = dict(semiconductor); p.setdefault('label', 'custom')
    else:
        assert semiconductor in CONTACT, f"'{semiconductor}' is not in CONTACT {list(CONTACT)}; pass a dict of cited values instead."
        p = dict(CONTACT[semiconductor]); p['label'] = semiconductor
    p.setdefault('Astar', 120.0 * p['m'])
    return p


def schottky_mott(metal, semiconductor, verbose=True):
    """Ideal (unpinned) barrier heights of a metal on a semiconductor: Phi_Bn = W_m - chi, Phi_Bp = Eg - Phi_Bn.
    metal: a WORK_FUNCTION key or a work function in eV."""
    p = _semi(semiconductor)
    Wm = WORK_FUNCTION[metal] if isinstance(metal, str) else float(metal)
    phin = Wm - p['chi']; phip = p['Eg'] - phin
    if verbose:
        print(f"{metal} (W_m = {Wm:.2f} eV) on {p['label']}: ideal Phi_Bn = {phin:+.2f} eV, Phi_Bp = {phip:+.2f} eV "
              f"(real interfaces are pinned — compare with the measured barrier)")
    return phin, phip


def E00(N, semiconductor):
    """Tunneling characteristic energy (eV) for a doping N (cm^-3) under the metal (Padovani & Stratton 1966)."""
    p = _semi(semiconductor)
    return (q_C * hbar / 2) * np.sqrt(np.asarray(N, float) * 1e6 / (p['m'] * m0 * p['eps_r'] * eps0 * 100)) / q_C


def contact_resistivity(N, phiB, semiconductor, T=300.0, verbose=True):
    """Specific contact resistivity (ohm cm^2) of a metal–semiconductor contact with barrier phiB (eV) and doping N
    (cm^-3) under the metal: rho_c = (k / (q A* T)) exp(phiB / E0), E0 = E00 coth(E00/kT). This interpolates between
    thermionic emission (E00 << kT) and field emission (E00 >> kT) (Yu 1970; Schroder 2006, Ch. 3)."""
    p = _semi(semiconductor); kT_ = kB_eV * T
    e00 = E00(N, p); e0 = e00 / np.tanh(e00 / kT_)
    rc = kB_eV / (p['Astar'] * T) * np.exp(phiB / e0)
    if verbose and np.ndim(N) == 0:
        r = e00 / kT_
        regime = 'thermionic emission' if r < 0.5 else ('thermionic-field emission' if r < 2.0 else 'field emission (tunneling)')
        print(f"{p['label']}, N = {N:.1e} cm^-3, Phi_B = {phiB:.2f} eV: E00 = {1e3*e00:.1f} meV = {r:.2f} kT -> {regime}; "
              f"rho_c = {rc:.2e} ohm cm^2")
    return rc


def rhoc_vs_doping(semiconductor, barriers=(0.2, 0.4, 0.6, 0.8), N=None, target=None, reported=(), ax=None):
    """rho_c against the doping under the metal for several barrier heights.
    target: (low, high) band of rho_c your device needs (ohm cm^2). reported: [(N, rho_c, label), ...] from your corpus."""
    p = _semi(semiconductor)
    N = np.logspace(16, 20.7, 300) if N is None else np.asarray(N, float)
    if ax is None:
        _, ax = plt.subplots(figsize=(6.8, 4.5))
    for phiB in barriers:
        ax.loglog(N, contact_resistivity(N, phiB, p, verbose=False), lw=1.8, label=rf'$\Phi_B$ = {phiB:.2f} eV')
    if target is not None:
        ax.axhspan(target[0], target[1], color='0.6', alpha=0.25, label='target for your device')
    for Nr, rr, lab in reported:
        ax.plot(Nr, rr, 'o', ms=7, label=lab)
    ax.set_ylim(1e-8, 1e4)
    ax.set_xlabel(r'Doping Under the Metal (cm$^{-3}$)'); ax.set_ylabel(r'Specific Contact Resistivity $\rho_c$ ($\Omega$ cm$^2$)')
    ax.set_title(f"Metal on {p['label']} at 300 K (TE–TFE–FE interpolation)", fontsize=10); ax.legend(fontsize=7)
    return ax


def contact_drop(rho_c, J, f_contact=1.0):
    """Voltage drop (V) across a contact of specific resistivity rho_c (ohm cm^2) for a device current density J
    (A cm^-2 of device area) when the contact covers the fraction f_contact of the device area."""
    dV = rho_c * J / f_contact
    print(f"rho_c = {rho_c:.1e} ohm cm^2, J = {J:g} A/cm^2, contact fraction {100*f_contact:.1f} %: drop = {1e3*dV:.2f} mV "
          f"(equivalent R_S = {rho_c/f_contact:.3g} ohm cm^2)")
    return dV


def tlm_fit(d_um, RT_ohm, W_cm, L_um=None, ax=None):
    """Transfer length method. d_um: pad spacings (µm); RT_ohm: resistance between adjacent pads (ohm); W_cm: pad width
    (cm). Fits R_T = 2 R_c + R_sh d / W and returns R_sh (ohm/sq), R_c (ohm), L_T (µm), rho_c (ohm cm^2).
    If the pad length L_um is given, rho_c is solved from R_c = (R_sh L_T / W) coth(L / L_T) instead of the long-contact limit."""
    d = np.asarray(d_um, float) * 1e-4; RT = np.asarray(RT_ohm, float)
    slope, icpt = np.polyfit(d, RT, 1)
    Rsh = slope * W_cm; Rc = icpt / 2
    if Rc <= 0:
        print('The intercept is not positive: the contact resistance is below the resolution of this data set.'); return Rsh, Rc, np.nan, np.nan
    LT = Rc * W_cm / Rsh
    if L_um is not None:
        L = L_um * 1e-4; lo, hi = 1e-7, 10.0
        for _ in range(200):                       # bisection on L_T: R_c = Rsh*LT/W*coth(L/LT) increases with LT
            mid = np.sqrt(lo * hi)
            if Rsh * mid / W_cm / np.tanh(L / mid) > Rc: hi = mid
            else: lo = mid
        LT = np.sqrt(lo * hi)
    rc = Rsh * LT ** 2
    resid = RT - (icpt + slope * d)
    print(f"TLM fit: R_sh = {Rsh:.1f} ohm/sq, R_c = {Rc:.3g} ohm, L_T = {1e4*LT:.1f} µm, rho_c = {rc:.2e} ohm cm^2 "
          f"(rms residual {np.sqrt(np.mean(resid**2)):.2g} ohm; smallest spacing = {d.min()/LT:.1f} L_T)")
    if ax is not None:
        x = np.linspace(-2.5 * LT, d.max() * 1.05, 50)
        ax.plot(1e4 * x, icpt + slope * x, lw=1.5); ax.plot(1e4 * d, RT, 'o', ms=6)
        ax.axvline(0, color='0.5', lw=0.8); ax.set_ylim(bottom=0)
        ax.set_xlabel('Pad Spacing $d$ (µm)'); ax.set_ylabel(r'Total Resistance $R_T$ ($\Omega$)')
    return Rsh, Rc, 1e4 * LT, rc


def grid_losses(s_mm, Rsh=100.0, wf_um=30.0, hf_um=12.0, rho_f_uohmcm=3.0, Lf_mm=5.7, rhoc_mohmcm2=2.0, Jmp_mAcm2=40.0, Vmp=0.62):
    """Fractional power losses of a front grid of parallel fingers at spacing s (Meier & Schroder 1984):
    shading w_f/s; emitter R_sh s^2 J/(12 V); finger rho_f L_f^2 s J/(3 w_f h_f V); contact rho_c s J/(w_f V)."""
    s = np.asarray(s_mm, float) * 0.1; wf, hf, Lf = wf_um * 1e-4, hf_um * 1e-4, Lf_mm * 0.1
    JV = Jmp_mAcm2 * 1e-3 / Vmp
    return dict(shading=wf / s, emitter=Rsh * s ** 2 * JV / 12, finger=rho_f_uohmcm * 1e-6 * Lf ** 2 * s * JV / (3 * wf * hf),
                contact=rhoc_mohmcm2 * 1e-3 * s * JV / wf)


def grid_optimization(s_mm=None, ax=None, **grid):
    """Plots the four grid losses and their total against finger spacing, finds the optimum spacing and reports the
    equivalent series resistance of the three resistive terms there. Keyword arguments as in grid_losses."""
    s = np.linspace(0.3, 4.0, 800) if s_mm is None else np.asarray(s_mm, float)
    L = grid_losses(s, **grid); tot = sum(L.values()); i = int(np.argmin(tot))
    Jmp = grid.get('Jmp_mAcm2', 40.0) * 1e-3; Vmp = grid.get('Vmp', 0.62)
    res = sum(v[i] for k, v in L.items() if k != 'shading'); Rs = res * Vmp / Jmp
    if ax is None:
        _, ax = plt.subplots(figsize=(6.8, 4.5))
    for (k, v), ls in zip(L.items(), ('--', '-', '-.', ':')):
        ax.plot(s, 100 * v, ls, lw=1.5, label=k)
    ax.plot(s, 100 * tot, 'k', lw=2.2, label='total'); ax.plot(s[i], 100 * tot[i], 'ko', ms=6)
    ax.set_ylim(0, min(15, 100 * tot.max())); ax.set_xlabel('Finger Spacing $s$ (mm)'); ax.set_ylabel('Fractional Power Loss (%)'); ax.legend(fontsize=8)
    print(f"Optimum finger spacing {s[i]:.2f} mm: " + ', '.join(f"{k} {100*v[i]:.2f} %" for k, v in L.items()) +
          f"; total {100*tot[i]:.2f} %. Resistive part = {100*res:.2f} % -> equivalent R_S = {Rs:.3f} ohm cm^2.")
    return dict(s_opt_mm=float(s[i]), total=float(tot[i]), Rs_grid=float(Rs), losses={k: float(v[i]) for k, v in L.items()})


def ff_with_rs(Rs_ohmcm2, Jsc_mAcm2, Voc, n=1.0, T=300.0, verbose=True):
    """Fill factor of a solar cell with series resistance R_S (Green, Solid-State Electron. 24, 788 (1981)):
    FF0 = (v - ln(v + 0.72))/(v + 1), v = Voc/(n kT/q); FF = FF0 (1 - 1.1 r_s) + r_s^2/5.4, r_s = R_S Jsc/Voc."""
    v = Voc / (n * kB_eV * T); FF0 = (v - np.log(v + 0.72)) / (v + 1)
    rs = np.asarray(Rs_ohmcm2, float) * Jsc_mAcm2 * 1e-3 / Voc
    FF = FF0 * (1 - 1.1 * rs) + rs ** 2 / 5.4
    if verbose and np.ndim(Rs_ohmcm2) == 0:
        print(f"R_S = {Rs_ohmcm2:.3g} ohm cm^2: r_s = {rs:.4f}, FF0 = {100*FF0:.2f} %, FF = {100*FF:.2f} % "
              f"({100*(FF0-FF):.2f} points lost); efficiency = {Voc*Jsc_mAcm2*FF:.2f} % against {Voc*Jsc_mAcm2*FF0:.2f} % (AM1.5G, 100 mW/cm^2)")
    return FF


def tco_sheet(n_cm3, mu_cm2Vs, t_nm, verbose=True):
    """Sheet resistance (ohm/sq) of a transparent conductor: R_sh = 1/(q n mu t)."""
    Rsh = 1.0 / (q_C * n_cm3 * mu_cm2Vs * t_nm * 1e-7)
    if verbose:
        print(f"n = {n_cm3:.1e} cm^-3, mu = {mu_cm2Vs:g} cm^2/Vs, t = {t_nm:g} nm: rho = {1/(q_C*n_cm3*mu_cm2Vs):.2e} ohm cm, R_sh = {Rsh:.1f} ohm/sq")
    return Rsh


def spreading_length(t_um, rho_ohmcm, J0_Acm2, n_ideal=1.5, verbose=True):
    """Current spreading length (µm) of an LED top layer of thickness t and resistivity rho at a current density J0
    under the contact edge: L_s = sqrt(t n kT / (q rho J0)) (Thompson 1980; Schubert 2018)."""
    Ls = 1e4 * np.sqrt(t_um * 1e-4 * n_ideal * kT / (rho_ohmcm * J0_Acm2))
    if verbose:
        print(f"t = {t_um:g} µm, rho = {rho_ohmcm:g} ohm cm (R_sh = {rho_ohmcm/(t_um*1e-4):.0f} ohm/sq), J0 = {J0_Acm2:g} A/cm^2: "
              f"L_s = {Ls:.1f} µm -> contact fingers no more than about {2*Ls:.0f} µm apart")
    return Ls


def photodiode_bandwidth(diameter_um, w_um, eps_r, Rs_ohm=5.0, RL_ohm=50.0, vsat_cms=1e7, C_pad_fF=0.0, verbose=True):
    """3 dB bandwidth (GHz) of a p-i-n photodiode of the given diameter and depletion width w:
    f_RC = 1/(2 pi (R_S + R_L)(C_j + C_pad)), C_j = eps A / w; f_tr = 0.45 v_sat / w; 1/f^2 = 1/f_RC^2 + 1/f_tr^2."""
    A = np.pi * (diameter_um * 1e-4 / 2) ** 2; w = np.asarray(w_um, float) * 1e-4
    Cj = eps_r * eps0 * A / w; C = Cj + C_pad_fF * 1e-15
    fRC = 1 / (2 * np.pi * (Rs_ohm + RL_ohm) * C); ftr = 0.45 * vsat_cms / w
    f = 1 / np.sqrt(1 / fRC ** 2 + 1 / ftr ** 2)
    if verbose and np.ndim(w_um) == 0:
        print(f"diameter {diameter_um:g} µm, w = {w_um:g} µm: C_j = {1e15*Cj:.0f} fF, f_RC = {fRC/1e9:.1f} GHz, f_transit = {ftr/1e9:.1f} GHz, "
              f"f_3dB = {f/1e9:.1f} GHz (R_S = {Rs_ohm:g} ohm contributes {100*Rs_ohm/(Rs_ohm+RL_ohm):.0f} % of the RC time)")
    return f / 1e9

print('Module A loaded: schottky_mott, E00, contact_resistivity, rhoc_vs_doping, contact_drop, tlm_fit, grid_losses, '
      'grid_optimization, ff_with_rs, tco_sheet, spreading_length, photodiode_bandwidth. Tables: CONTACT, WORK_FUNCTION.')
```

### Background: from one device to a part with terminals

A single junction delivers (or needs) about one volt, a current set by its area, and has no terminals a user can reach. Interconnection joins devices in series and parallel and brings the current out: tabbing ribbons, round wires, conductive adhesive or monolithic laser-scribed interconnects for solar cells; wire bonds, flip-chip bumps and die attach for LED, laser and detector chips. In lecture this was the second of four steps of *final device integration* — singulate and mount, interconnect, package and encapsulate, test and bin — and three quantities came out of it:

* **Joule loss and shading in cell interconnects.** Current builds up linearly along a wire soldered across a cell, so $P_{loss} = \tfrac{2}{3} I^2 \rho L/(nA)$ for $n$ conductors of cross-section $A$ and length $L$; as a fraction of cell power it scales with $IL/(nA)$, which is why cells are cut in half and carry many thin wires (Braun et al., Energy Procedia 38, 334 (2013)). All the gains and losses of integration are collected in the cell-to-module (CTM) ratio (Hädrich et al., Sol. Energy Mater. Sol. Cells 131, 14 (2014)).
* **Mismatch.** Devices in series share a current, so the weakest one limits the string and can be driven into reverse bias, where it dissipates power instead of generating it; bypass diodes limit both the power loss and the reverse voltage (Bishop, Solar Cells 25, 73 (1988)). LED chips in parallel have the mirror-image problem: they share a voltage, and the hottest chip takes the most current.
* **Parasitics and stress.** A bond wire is an inductor of roughly 1 nH per millimeter, and with the junction capacitance it makes the detector a second-order circuit, $H(f) = [1 - (2\pi f)^2 L C_j + j 2\pi f R C_j]^{-1}$. Every joint also connects materials with different thermal expansion, and the mismatch strain $\varepsilon = \Delta\alpha\,\Delta T$ accumulated on cooling from the joining temperature and in every thermal cycle is what fatigues solder joints and cracks cells (Frear, Ch. 53 in Kasap & Capper, *Springer Handbook of Electronic and Photonic Materials*, 2nd ed. (2017)).

```python
# ============================================================
# HW04 MODULE B — interconnection and packaging calculators: interconnect Joule loss and shading, series strings with
# mismatch and bypass diodes, bond-wire parasitics, thermomechanical strain, escape cone, thermal resistance,
# module temperature, moisture ingress
# (instructor-provided; run, do not modify)
# ============================================================
import numpy as np, pandas as pd
import matplotlib.pyplot as plt

# Thermal conductivity k (W m^-1 K^-1) and linear thermal expansion coefficient alpha (ppm/K) at room temperature.
# Representative handbook and datasheet values (metals, ceramics and Sn–Pb after Kasap & Capper 2017, Table 53.3).
# Starting points only — use the values of the materials in your corpus and cite them. None = not tabulated here.
PACKAGE_MATERIALS = {
    'Si': (150, 2.6), 'GaAs': (55, 5.7), 'GaN': (180, 5.6), 'InP': (68, 4.6), 'sapphire': (35, 5.3),
    'Cu': (400, 16.8), 'Al': (240, 23.5), 'Ag': (430, 19.7), 'Au': (315, 14.2), 'Kovar': (17, 5.5),
    'AlN': (180, 4.3), 'alumina': (25, 6.4), 'soda-lime glass': (1.0, 9.0),
    'AuSn solder': (57, 16.0), 'SAC305 solder': (58, 22.0), 'SnPb solder': (50, 25.0), 'SnBi solder': (19, 15.0),
    'sintered Ag': (200, 19.0), 'Ag epoxy': (3.0, 50.0), 'thermal grease': (3.0, None),
    'silicone': (0.2, 300.0), 'epoxy': (0.2, 60.0), 'EVA': (0.3, 270.0), 'FR-4': (0.3, 15.0), 'MCPCB dielectric': (2.0, None),
}


def interconnect_loss(I_A, L_mm, n, d_mm=None, w_mm=None, t_mm=None, rho_uohmcm=1.72, Vmp=0.62, cell_width_mm=182.0,
                      optical_factor=None, verbose=True):
    """Joule loss in the interconnects of one solar cell, P = (2/3) I^2 rho L / (n A), for n round wires of diameter
    d_mm or n flat ribbons w_mm x t_mm running the length L_mm of the cell, with the geometric shading n*width/cell width.
    optical_factor: effective optical width / geometric width (about 0.65 for round wires in a module, 1 for ribbons)."""
    if d_mm is not None:
        A = np.pi * (d_mm / 20) ** 2; width = d_mm; kind = f'{n} wires, {d_mm:g} mm'
        optical_factor = 0.65 if optical_factor is None else optical_factor
    else:
        A = (w_mm / 10) * (t_mm / 10); width = w_mm; kind = f'{n} ribbons, {w_mm:g} x {t_mm:g} mm'
        optical_factor = 1.0 if optical_factor is None else optical_factor
    P = 2 / 3 * I_A ** 2 * rho_uohmcm * 1e-6 * (L_mm / 10) / (n * A)
    frac = P / (I_A * Vmp); shade = n * width / cell_width_mm
    if verbose:
        print(f"{kind}; I = {I_A:g} A, L = {L_mm:g} mm: Joule loss {P:.3f} W = {100*frac:.2f} % of cell power; "
              f"shading {100*shade:.2f} % geometric, {100*shade*optical_factor:.2f} % effective; sum {100*(frac + shade*optical_factor):.2f} %")
    return dict(P_W=P, joule=frac, shading=shade, shading_eff=shade * optical_factor)


def _cell_curve(IL, I0, Vt, Rs, Rsh, Vbr, a=2.0e-3, m=3.0):
    """Single-diode cell with the Bishop reverse-breakdown term (Bishop, Solar Cells 25, 73 (1988)); returns I, V sorted by I."""
    vj = np.concatenate([np.linspace(Vbr + 1e-3, Vbr + 1.0, 4000), np.linspace(Vbr + 1.0, 1.2 * Vt * np.log(IL / I0 + 1), 20000)[1:]])
    i = IL - I0 * (np.exp(vj / Vt) - 1.0) - vj / Rsh * (1.0 + a * (1.0 - vj / Vbr) ** (-m))
    v = vj - i * Rs
    o = np.argsort(i)
    return i[o], v[o]


def string_iv(n_cells=72, n_substrings=3, shaded=None, IL=6.9, Voc=0.72, Rs=3.0e-3, Rsh=100.0, Vbr=-25.0, Vd=0.5,
              T=298.15, ax=None, verbose=True):
    """I–V and P–V curves of a series string of n_cells identical cells in n_substrings equal substrings, each with a
    bypass diode of forward drop Vd, with and without the diodes. shaded: {cell index (0-based): fraction of full
    photocurrent}, e.g. {30: 0.4}. IL, Voc: photocurrent (A) and open-circuit voltage (V) of one unshaded cell;
    Rs, Rsh in ohm; Vbr: reverse breakdown voltage of a cell (V, negative). Returns the three maximum powers."""
    shaded = shaded or {}
    Vt = 8.617333e-5 * T; I0 = IL / (np.exp(Voc / Vt) - 1.0)
    I = np.linspace(0.0, 1.05 * IL, 3000)
    def vcell(f):
        i, v = _cell_curve(f * IL, I0, Vt, Rs, Rsh, Vbr)
        return np.interp(I, i, v)
    v_full = vcell(1.0); per = n_cells // n_substrings
    subs, vsh = [], {}
    for k in range(n_substrings):
        v = np.zeros_like(I)
        for c in range(k * per, (k + 1) * per):
            if c in shaded:
                vsh[c] = vcell(shaded[c]); v = v + vsh[c]
            else:
                v = v + v_full
        subs.append(v)
    V_ok = n_cells * v_full; V_nobp = sum(subs); V_bp = sum(np.maximum(s, -Vd) for s in subs)
    pmax = lambda V: float(np.max(np.where(V >= 0, V * I, 0)))
    out = dict(P_matched=pmax(V_ok), P_bypass=pmax(V_bp), P_no_bypass=pmax(V_nobp))
    if ax is None:
        _, ax = plt.subplots(1, 2, figsize=(10, 4))
    for V, lab, ls in ((V_ok, 'all cells matched', 'k-'), (V_nobp, 'shaded, no bypass diodes', 'r--'), (V_bp, 'shaded, with bypass diodes', 'b-')):
        ax[0].plot(V, I, ls, lw=1.8, label=lab); ax[1].plot(V, V * I, ls, lw=1.8, label=lab)
    ax[0].set_xlim(0, 1.05 * n_cells * Voc); ax[0].set_ylim(0, 1.15 * IL); ax[1].set_xlim(0, 1.05 * n_cells * Voc); ax[1].set_ylim(0, 1.1 * out['P_matched'])
    ax[0].set_xlabel('String Voltage (V)'); ax[0].set_ylabel('String Current (A)'); ax[1].set_xlabel('String Voltage (V)'); ax[1].set_ylabel('String Power (W)')
    ax[0].legend(fontsize=8, loc='lower left')
    if verbose:
        print(f"P_max: matched {out['P_matched']:.0f} W; with bypass diodes {out['P_bypass']:.0f} W ({100*out['P_bypass']/out['P_matched']:.0f} %); "
              f"without {out['P_no_bypass']:.0f} W ({100*out['P_no_bypass']/out['P_matched']:.0f} %)")
        for c, v in vsh.items():
            k = c // per; j = int(np.argmin(np.abs(subs[k] + Vd)))       # bypass turn-on of that substring
            j0 = int(np.argmin(np.abs(V_nobp)))                           # string short-circuited, no diodes
            print(f"   shaded cell {c} ({100*shaded[c]:.0f} % of full current): with diodes, at bypass turn-on V_cell = {v[j]:.1f} V, "
                  f"dissipation {-v[j]*I[j]:.0f} W; without diodes at string short circuit V_cell = {v[j0]:.1f} V, dissipation {-v[j0]*I[j0]:.0f} W")
    return out


def monolithic_strip(Rsh_tco, dead_um=250.0, Jmp_mAcm2=30.0, Vmp=0.70, w_mm=None, ax=None):
    """Monolithically interconnected thin-film module: fractional loss against the active width w of a cell strip.
    TCO sheet loss R_sh w^2 J_mp / (3 V_mp) (all current exits at one edge) plus the dead-area fraction of the
    P1–P3 scribe zone, dead/(w + dead). Returns the optimum active width (mm) and the total loss there."""
    w = np.linspace(1.0, 20.0, 800) if w_mm is None else np.asarray(w_mm, float)
    p_tco = Rsh_tco * (w * 0.1) ** 2 * Jmp_mAcm2 * 1e-3 / (3 * Vmp); p_dead = dead_um * 1e-3 / (w + dead_um * 1e-3)
    tot = p_tco + p_dead; i = int(np.argmin(tot))
    print(f"R_sh(TCO) = {Rsh_tco:g} ohm/sq, scribe zone {dead_um:g} µm: optimum active strip width {w[i]:.1f} mm; "
          f"TCO loss {100*p_tco[i]:.2f} %, dead area {100*p_dead[i]:.2f} %, total {100*tot[i]:.2f} %")
    if ax is not None:
        ax.plot(w, 100 * p_tco, '-', lw=1.5, label='TCO sheet resistance'); ax.plot(w, 100 * p_dead, '--', lw=1.5, label='dead area (scribes)')
        ax.plot(w, 100 * tot, 'k', lw=2.2, label='total'); ax.plot(w[i], 100 * tot[i], 'ko', ms=6)
        ax.set_ylim(0, min(25, 100 * tot.max())); ax.set_xlabel('Active Strip Width (mm)'); ax.set_ylabel('Fractional Power Loss (%)'); ax.legend(fontsize=8)
    return float(w[i]), float(tot[i])


def bondwire_response(Cj_pF, L_nH, R_ohm=50.0, f_GHz=None, ax=None, label=None):
    """Frequency response of a photodiode (junction capacitance C_j) connected to a load R through a bond wire of
    inductance L: H(f) = 1 / (1 - (2 pi f)^2 L C_j + j 2 pi f R C_j). Returns the 3 dB bandwidth in GHz and prints
    the peaking. L of a bond wire is roughly 1 nH per mm."""
    f = np.logspace(8, 11.5, 4000) if f_GHz is None else np.asarray(f_GHz, float) * 1e9
    w = 2 * np.pi * f; C = Cj_pF * 1e-12; L = L_nH * 1e-9
    dB = 20 * np.log10(np.abs(1.0 / (1.0 - w ** 2 * L * C + 1j * w * R_ohm * C)))
    below = np.where(dB < -3.0)[0]; f3 = f[below[0]] / 1e9 if len(below) else np.nan
    print(f"C_j = {Cj_pF:g} pF, L = {L_nH:g} nH, R = {R_ohm:g} ohm: f_3dB = {f3:.1f} GHz (RC limit {1/(2*np.pi*R_ohm*C)/1e9:.1f} GHz), peaking {max(dB.max(), 0):.1f} dB")
    if ax is not None:
        ax.semilogx(f / 1e9, dB, lw=1.8, label=label or f'{L_nH:g} nH: {f3:.0f} GHz'); ax.axhline(-3, color='0.5', lw=0.8, ls=':')
        ax.set_ylim(-12, 8); ax.set_xlabel('Frequency (GHz)'); ax.set_ylabel('Response (dB)'); ax.legend(fontsize=8)
    return f3


def mismatch_strain(material_1, material_2, T_join_C, T_use_C=25.0, cycle=(-40.0, 85.0)):
    """Thermal-expansion mismatch strain between two joined materials cooled from the joining temperature, and the
    strain range of one thermal cycle: epsilon = (alpha_1 - alpha_2) dT. Materials: PACKAGE_MATERIALS keys or alpha in ppm/K."""
    a = [PACKAGE_MATERIALS[m][1] if isinstance(m, str) else float(m) for m in (material_1, material_2)]
    assert None not in a, 'No expansion coefficient tabulated for one of the materials: pass alpha in ppm/K instead.'
    da = abs(a[0] - a[1]) * 1e-6; e_join = da * (T_join_C - T_use_C); e_cyc = da * (cycle[1] - cycle[0])
    print(f"{material_1} ({a[0]:g} ppm/K) on {material_2} ({a[1]:g} ppm/K): mismatch {1e6*da:.1f} ppm/K; strain after cooling from "
          f"{T_join_C:g} to {T_use_C:g} °C = {100*e_join:.3f} %; per {cycle[0]:g}/{cycle[1]:+g} °C cycle = {100*e_cyc:.3f} %")
    return e_join, e_cyc


def escape_fraction(n_semi, n_enc=1.0, fresnel=True, verbose=True):
    """Escape cone of light generated in a semiconductor of index n_semi leaving a flat face into a medium n_enc:
    theta_c = asin(n_enc/n_semi); fraction of isotropic emission inside the cone of one face = (1 - cos theta_c)/2,
    multiplied by the normal-incidence Fresnel transmittance if fresnel=True."""
    th = np.arcsin(min(1.0, n_enc / n_semi)); eta = 0.5 * (1 - np.cos(th))
    T = 1 - ((n_semi - n_enc) / (n_semi + n_enc)) ** 2 if fresnel else 1.0
    if verbose:
        print(f"n = {n_semi:g} into n = {n_enc:g}: critical angle {np.degrees(th):.1f}°, fraction inside the cone {100*eta:.2f} % per face"
              + (f", {100*eta*T:.2f} % after Fresnel transmission ({100*T:.1f} %)" if fresnel else ''))
    return eta * T


def thermal_stack(layers, P_heat_W, T_ambient_C=25.0, extra_K_per_W=0.0, verbose=True):
    """Junction temperature from a series stack of layers: R_th = t/(k A) for each, T_j = T_a + P_heat * sum(R_th).
    layers: [(name, thickness_um, k or PACKAGE_MATERIALS key, area_mm2), ...]; extra_K_per_W: the rest of the path
    (heat sink to air, or a datasheet junction-to-case value) given directly."""
    rows = []
    for name, t_um, k, A_mm2 in layers:
        kk = PACKAGE_MATERIALS[k][0] if isinstance(k, str) else float(k)
        rows.append(dict(layer=name, t_um=t_um, k_WmK=kk, A_mm2=A_mm2, Rth_KperW=t_um * 1e-6 / (kk * A_mm2 * 1e-6)))
    if extra_K_per_W:
        rows.append(dict(layer='rest of path (given)', t_um=np.nan, k_WmK=np.nan, A_mm2=np.nan, Rth_KperW=extra_K_per_W))
    df = pd.DataFrame(rows); Rth = df.Rth_KperW.sum(); Tj = T_ambient_C + P_heat_W * Rth
    df['dT_K'] = df.Rth_KperW * P_heat_W
    if verbose:
        print(df.round(3).to_string(index=False))
        print(f"Total R_th = {Rth:.2f} K/W; P_heat = {P_heat_W:g} W -> T_j = {Tj:.1f} °C at {T_ambient_C:g} °C ambient")
    return Tj, df


def module_temperature(T_ambient_C, G_Wm2=1000.0, NMOT_C=43.0, gamma_pct_per_K=-0.30, verbose=True):
    """Operating temperature of a PV module from its nominal module operating temperature,
    T_mod = T_a + (NMOT - 20) G/800, and the power relative to the 25 °C rating for a temperature coefficient gamma."""
    Tm = T_ambient_C + (NMOT_C - 20.0) * G_Wm2 / 800.0; dP = gamma_pct_per_K * (Tm - 25.0)
    if verbose:
        print(f"T_a = {T_ambient_C:g} °C, G = {G_Wm2:g} W/m^2, NMOT = {NMOT_C:g} °C: T_mod = {Tm:.1f} °C; power {dP:+.1f} % relative to the 25 °C rating "
              f"(gamma = {gamma_pct_per_K:g} %/K)")
    return Tm, dP


def moisture_half_time(WVTR_gm2day, Csat_l_gm2=1.0, verbose=True):
    """Time for the encapsulant behind a barrier of water-vapor transmission rate WVTR (g m^-2 day^-1) to reach half
    of its saturated water content C_sat*l (g/m^2; about 1 g/m^2 for 0.46 mm of EVA): tau = 0.693 C_sat l / WVTR
    (after Dameron et al. 2008). Returns days."""
    tau = 0.693 * Csat_l_gm2 / np.asarray(WVTR_gm2day, float)
    if verbose and np.ndim(WVTR_gm2day) == 0:
        print(f"WVTR = {WVTR_gm2day:.1e} g/m^2/day: half-time = {tau:.3g} days = {tau/365:.3g} years")
    return tau

print('Module B loaded: interconnect_loss, string_iv, monolithic_strip, bondwire_response, mismatch_strain, escape_fraction, thermal_stack, '
      'module_temperature, moisture_half_time. Table: PACKAGE_MATERIALS (k, alpha).')
```

### Background: four functions of a package

A package is judged by four functions — electrical, optical, thermal and protective — and the priorities differ by device.

* **Optical.** An encapsulant of index $n_{enc}$ between the semiconductor and air widens the escape cone of an emitter, $\theta_c = \sin^{-1}(n_{enc}/n_s)$, with a fraction $\tfrac{1}{2}(1 - \cos\theta_c) \approx n_{enc}^2/(4n_s^2)$ of the isotropic emission inside the cone of one face, provided the outer surface is shaped (a dome) so that the light then leaves the encapsulant (Schubert (2018), Ch. 5 and 11). For an absorber the same index matching removes the reflection at the glass–cell air gap and changes the optimum antireflection coating (Homework 3, Task 4).
* **Thermal.** Heat leaves by conduction through the package: each layer contributes $R_{th} = t/(kA)$, resistances in series add, and $T_j = T_a + R_{th}P_{heat}$. The layers directly under a small chip dominate, because the heat flux is highest there. A PV module has no heat sink; its temperature follows from its nominal module operating temperature, $T_{mod} \approx T_a + (\mathrm{NMOT} - 20\,{}^{\circ}\mathrm{C})\,G/(800\ \mathrm{W\,m^{-2}})$, and its power from its temperature coefficient.
* **Protective.** Water and oxygen are kept out by a barrier characterized by its water-vapor transmission rate (WVTR); the time for the encapsulant behind it to half-saturate is roughly $\tau_{1/2} = 0.693\,C_{sat}\,l/\mathrm{WVTR}$ (Kempe, Sol. Energy Mater. Sol. Cells 90, 2720 (2006); Dameron et al., Photovoltaics International, 5th ed., 121 (2009)). A polymer backsheet is not a barrier on the scale of a product lifetime; glass–glass with an edge seal, or a hermetic metal or ceramic package, is.
* **Electrical.** Insulation, terminals, bypass or electrostatic-discharge protection diodes, and controlled impedance for high-speed parts — treated in Task 2.

Qualification tests are how a package design is accepted: IEC 61215 (thermal cycling, damp heat, humidity–freeze, mechanical load, hot-spot endurance) for PV modules, IES LM-80 with TM-21 for the lumen maintenance of LED packages, and Telcordia GR-468 for telecom optoelectronics.

## Appendix B — Methods Used in Calculators

**B.1 Contact resistivity across the transport regimes (Padovani & Stratton 1966; Yu 1970; Schroder 2006).** For thermionic emission over a barrier $\Phi_B$ the current is $J = A^{*}T^2 e^{-\Phi_B/kT}(e^{qV/kT} - 1)$ and its slope at zero bias gives $\rho_c = (k/qA^{*}T)\,e^{\Phi_B/kT}$. When the depletion layer is thin enough to tunnel through, the exponent's energy scale $kT$ is replaced by $E_{00} = (q\hbar/2)\sqrt{N/(m^{*}\varepsilon_s)}$; `contact_resistivity` uses the interpolation $E_0 = E_{00}\coth(E_{00}/kT)$, which reduces to $kT$ for $E_{00} \ll kT$ and to $E_{00}$ for $E_{00} \gg kT$, with the thermionic prefactor throughout. The exact tunneling prefactors differ from it by factors of order unity to ten, so read the result as the correct *trend* with doping and barrier and the correct order of magnitude, not as a prediction to two figures. The regime labels use $E_{00}/kT < 0.5$ (TE), 0.5–2 (TFE) and $> 2$ (FE).

**B.2 Transfer length method (Berger 1972).** Between two pads a distance $d$ apart on a stripe of width $W$, $R_T = 2R_c + R_{sh}d/W$. Under a pad of length $L$ the current transfers from the sheet to the metal over the transfer length $L_T = \sqrt{\rho_c/R_{sh}}$, and $R_c = (R_{sh}L_T/W)\coth(L/L_T)$. `tlm_fit` fits the straight line, takes $R_{sh}$ from the slope and $R_c$ from the intercept, and solves for $L_T$ (in the long-contact limit, $L_T = R_cW/R_{sh}$, unless you give the pad length) and $\rho_c = R_{sh}L_T^2$. It assumes that the sheet resistance under the pad equals that between pads and that the current is confined to the stripe; the printed ratio of the smallest spacing to $L_T$ tells you how far the intercept was extrapolated.

**B.3 Grid losses and fill factor (Meier & Schroder 1984; Green 1981).** For parallel fingers at spacing $s$ collecting a uniform current density, integrating $I^2R$ gives the fractional losses $p_{em} = R_{sh}s^2J_{mp}/(12V_{mp})$ in the sheet between fingers, $p_f = \rho_fL_f^2 sJ_{mp}/(3w_fh_fV_{mp})$ along the fingers, and $p_c = \rho_c sJ_{mp}/(w_fV_{mp})$ at the contact (taking the whole finger width as contact area, valid when $w_f \lesssim 2L_T$), against the shading $w_f/s$. `grid_optimization` minimizes their sum numerically. A fractional resistive loss $\ell = p_{em} + p_f + p_c$ corresponds to $R_S = \ell\,V_{mp}/J_{mp}$, and `ff_with_rs` uses Green's empirical expressions $\mathrm{FF}_0 = [v_{oc} - \ln(v_{oc} + 0.72)]/(v_{oc} + 1)$ with $v_{oc} = qV_{OC}/(n_{ideal}kT)$ and $\mathrm{FF} = \mathrm{FF}_0(1 - 1.1r_s) + r_s^2/5.4$. In a monolithically interconnected thin-film module all the current of a cell strip of active width $w$ leaves through the TCO at one edge, so the sheet term is $R_{sh}w^2J_{mp}/(3V_{mp})$ — four times the two-sided expression at the same width — and the competing loss is the dead area of the scribe zone, $w_{dead}/(w + w_{dead})$; `monolithic_strip` minimizes their sum.

**B.4 Series strings with mismatch (Bishop 1988).** `string_iv` describes each cell by a single-diode equation with series and shunt resistance and a reverse-breakdown term, $I = I_L - I_0(e^{V_j/V_T} - 1) - (V_j/R_{sh})[1 + a(1 - V_j/V_{br})^{-m}]$ with $V = V_j - IR_s$, $a = 2\times10^{-3}$ and $m = 3$. Because the cells are in series, the string voltage at each current is the sum of the cell voltages at that current; a substring with a bypass diode cannot go more negative than $-V_d$. The module reports the maximum power of the matched string and of the shaded string with and without bypass diodes, and the reverse voltage and dissipation of each shaded cell. The breakdown parameters are illustrative; the reverse characteristic of real cells varies widely with technology and from cell to cell.

**B.5 Package calculations.** Interconnect loss: the current in a conductor soldered along a cell rises linearly from zero to $I/n$, so the front conductors dissipate $I^2\rho L/(3nA)$ and the same again on the rear of the neighboring cell, $P_{loss} = \tfrac{2}{3}I^2\rho L/(nA)$. Bond wire: the photodiode is a current source with $C_j$ in parallel, feeding the load $R$ through the inductance $L$, which gives the second-order $H(f)$ of Task 2; the maximally flat response is at $L = R^2C_j/2$. Escape cone: the solid angle of a cone of half-angle $\theta_c$ is $2\pi(1 - \cos\theta_c)$ out of $4\pi$, hence $\tfrac{1}{2}(1 - \cos\theta_c)$ per face; `escape_fraction` multiplies by the normal-incidence Fresnel transmittance as a first estimate. Thermal stack: one-dimensional conduction with no heat spreading, which overestimates the resistance of layers much wider than the chip — give each layer the area the heat actually crosses. Moisture: $\tau_{1/2} = 0.693\,C_{sat}l/\mathrm{WVTR}$ treats the barrier as the only resistance to ingress and the encapsulant as a well-mixed reservoir.
